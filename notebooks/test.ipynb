```python
from pyspark.sql import SparkSession

# Start Spark Session
spark = SparkSession.builder \
    .appName("SparkTest") \
    .master("local[*]") \
    .getOrCreate()

print("Spark Session started")
print("Spark version:", spark.version)

# Create a simple DataFrame
data = [("Alice", 29), ("Bob", 31), ("Cathy", 25)]
df = spark.createDataFrame(data, ["Name", "Age"])

print("DataFrame:")
df.show()

# Run a simple transformation
df.filter(df.Age > 28).show()

spark.stop()
```

```python
import sparknlp
from sparknlp.base import DocumentAssembler
from sparknlp.annotator import Tokenizer, Normalizer
from pyspark.ml import Pipeline

# Start Spark NLP session
spark = sparknlp.start()
print("Spark NLP Session started")
print("Spark NLP version:", sparknlp.version())
print("Apache Spark version:", spark.version)

# Sample data
data = spark.createDataFrame([["Hello Spark NLP, this is a test!"]]).toDF("text")

# Document Assembler
document_assembler = DocumentAssembler() \
    .setInputCol("text") \
    .setOutputCol("document")

# Tokenizer
tokenizer = Tokenizer() \
    .setInputCols(["document"]) \
    .setOutputCol("token")

# Normalizer
normalizer = Normalizer() \
    .setInputCols(["token"]) \
    .setOutputCol("normalized")

# Build pipeline
pipeline = Pipeline(stages=[document_assembler, tokenizer, normalizer])

# Run pipeline
result = pipeline.fit(data).transform(data)
result.select("token.result", "normalized.result").show(truncate=False)
```
